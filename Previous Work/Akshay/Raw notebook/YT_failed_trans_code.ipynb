```python
import yt_dlp
import os
import re

# ==========================================================
# 1. ADD YOUR THREE LINKS HERE
# ==========================================================
video_links = [
    "https://www.youtube.com/watch?v=-URJ4t1q6ic", # Video 1
    "https://www.youtube.com/watch?v=1AO3UDuwEBQ", # Video 2
    "https://www.youtube.com/watch?v=1j8fCkoI2J0"  # Video 3
]

def clean_filename(title):
    return re.sub(r'[\\/*?:"<>|]', "", title).replace(" ", "_")

def clean_vtt_content(vtt_path):
    """Parses VTT files to clean out timestamps and formatting tags"""
    if not os.path.exists(vtt_path):
        return ""
    with open(vtt_path, 'r', encoding='utf-8') as f:
        lines = f.readlines()

    clean_lines = []
    for line in lines:
        line = line.strip()
        # Skip VTT headers, timestamps, and empty lines
        if (not line or
            line.startswith("WEBVTT") or
            line.startswith("Kind:") or
            line.startswith("Language:") or
            "-->" in line):
            continue
        # Strip internal XML formatting tags like <c> text </c>
        cleaned = re.sub(r'<[^>]*>', '', line)
        if cleaned and (not clean_lines or clean_lines[-1] != cleaned):
            clean_lines.append(cleaned)

    return " ".join(clean_lines)

# ==========================================================
# 2. PROCESS VIDEO CAPTIONS
# ==========================================================
for index, url in enumerate(video_links, start=1):
    print(f"\n[QUEUE] Processing Video {index} of {len(video_links)}")
    print(f"Target URL: {url}")

    # Subtitle download configuration bypassing the video stream blocks
    ydl_opts = {
        'writeautomaticsub': True,  # Pull auto-generated captions
        'subtitleslangs': ['hi', 'en'],  # Prefer Hindi, fallback to English
        'skip_download': True,      # Avoid downloading the actual video/audio file
        'outtmpl': f'temp_subs_{index}.%(ext)s',
        'quiet': True
    }

    try:
        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            print("-> Requesting transcript streams directly...")
            info_dict = ydl.extract_info(url, download=True)
            video_title = info_dict.get('title', f'Video_{index}')
            safe_title = clean_filename(video_title)

        # Target caption file candidates
        hi_subs = f'temp_subs_{index}.hi.vtt'
        en_subs = f'temp_subs_{index}.en.vtt'

        chosen_subs = hi_subs if os.path.exists(hi_subs) else (en_subs if os.path.exists(en_subs) else None)

        if chosen_subs:
            print(f"-> Extracting text data from payload...")
            transcript_text = clean_vtt_content(chosen_subs)

            output_filename = f"{safe_title}_transcript.txt"
            with open(output_filename, "w", encoding="utf-8") as f:
                f.write(transcript_text)

            print(f"✓ [SAVED] Transcript complete: '{output_filename}'")

            # Clean up the temporary structural file
            os.remove(chosen_subs)
            if chosen_subs == hi_subs and os.path.exists(en_subs):
                os.remove(en_subs)
        else:
            print(f"✗ [ERROR] YouTube did not return sub-tracks for this URL.")

    except Exception as e:
        print(f"✗ [CRASH] Pipeline failed: {e}")
        continue

print("\n" + "="*60 + "\nPipeline execution finished.")
```
