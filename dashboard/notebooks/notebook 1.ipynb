```python
#Date Imputation through Bands
import pandas as pd
import numpy as np

df2 = df
# Sample Data: Employee dates with some missing (NaT) or broken sequences

# 1. Ensure data is sorted by employee and date
df2 = df2.sort_values(['Test Date']).reset_index(drop=True)

# 2. Calculate the difference between consecutive dates
# We fill NaT temporarily with the previous value so it doesn't break the continuity check
temp_dates = df2['Test Date'].ffill()
date_diff = temp_dates.diff()

# 3. Define a boundary: A new window starts if the gap is greater than 1 day
# (or if it's the start of a new employee)
new_window_condition = (date_diff > pd.Timedelta(days=1))

# 4. Create the Window Index using cumsum()
df2['window_idx'] = new_window_condition.cumsum()

# 5. Impute the missing dates within each window
# Example: Forward fill, then backward fill within the same window group
df2['imputed_date'] = df2.groupby(['window_idx'])['Test Date'].ffill().bfill()

print(df2.head())
```

```python
import pandas as pd

# 1. Filter for dates starting from June 8, 2026
filtered_df = df[df['Test Date'] >= '2026-06-08']

# 2. Sort by the original DataFrame index to ensure chronological log order
filtered_df = filtered_df.sort_index()

# 3. Keep the 'first' occurrence of each date based on this index order
unique_dates_df = filtered_df.drop_duplicates(subset=['Test Date'], keep='first')

# 4. Extract sequential lists ordered by their appearance in the index
sequential_dates = unique_dates_df['Test Date'].dt.strftime('%d-%m-%Y').tolist()
original_indexes = unique_dates_df.index.tolist()

# Output the paired results
i = 0
for date, idx in zip(sequential_dates, original_indexes):
    print(f"{i}.Index: {idx} -> First Occurrence of Date: {date}")
    i+=1
```

```python
import pandas as pd

df5 = df

# # 1. Convert column to datetime (if not already done)
# df5['date'] = pd.to_datetime(df5['date'], dayfirst=True, errors='coerce')

# 2. Ensure data is sorted by its original index order before fixing
df5 = df5.sort_index()

# 3. Define the rules for a "Wrong" date based on your log patterns:
# - Rule A: The year skips into the future (e.g., greater than 2026)
# - Rule B: The date jumps ahead of our current real-world timeline
#           (Since today is in July 2026, logs from 2027+ or late 2026 are impossible)
current_year = 2026
is_wrong_year = df5['Test Date'].dt.year > current_year

# 4. Wipe out the incorrect dates by setting them to NaT (Not a Time)
df5.loc[is_wrong_year, 'Test Date'] = pd.NaT

# 5. Apply Strategy: Use the previous correct date to impute them
df5['Test Date'] = df5['Test Date'].ffill()

# (Optional) If any wrong dates appeared at the very first row,
# backfill just in case to avoid leaving a NaT at index 0
df5['Test Date'] = df5['Test Date'].bfill()
```

```python
import pandas as pd

df6 = df

# 1. Convert column to datetime
df6['Test Date'] = pd.to_datetime(df6['Test Date'], dayfirst=True, errors='coerce')

# 2. Ensure data is sorted by its original index order
df6 = df6.sort_index()

# 3. Define the threshold as today's actual date (July 14, 2026)
current_date = pd.to_datetime('2026-07-14')

# 4. Flag both future years (e.g., 2027+) and future months (e.g., August 2026)
is_impossible_future_date = df6['Test Date'] > current_date

# 5. Clear out the bad data rows
df6.loc[is_impossible_future_date, 'Test Date'] = pd.NaT

# 6. Apply your strategy: Forward fill from the last known good date
df6['Test Date'] = df6['Test Date'].ffill()

# Backfill as a safety net if the very first row happens to be broken
df6['Test Date'] = df6['Test Date'].bfill()
```

```python
# Normalize 'Question in Review Model?'
df6['Question in Review Model?'] = df6['Question in Review Model?'].replace({
    'Successfully Identified as Duplicate': 'Yes',
    'correctly identified as duplicate': 'Yes',
    'Wrongly Identified as Duplicate': 'No',
    'no': 'No',
    'yes': 'Yes'
}).fillna('Not Applicable')
df6['Question in Review Model?'] = df6['Question in Review Model?'].str.strip()

# Normalize 'Question Correctly Framed?'
df6['Question Correctly Framed?'] = df6['Question Correctly Framed?'].replace({
    'Well Framed': 'Yes',
    'Ambiguous': 'No',
    'Incorrectly Framed': 'No'
}).fillna('Not Applicable')
df6['Question Correctly Framed?'] = df6['Question Correctly Framed?'].str.strip()

# Normalize 'Original Language'
df6['Original Language'] = df6['Original Language'].replace({
    'ENGLISH': 'English',
    'ENGLISH ': 'English',
    'english': 'English',
    'Englsih': 'English',
    'ENG;ISH': 'English',
    'BENGALI': 'Bengali'
}).fillna('Not Specified')
df6['Original Language'] = df6['Original Language'].str.strip()

# Normalize 'Translated Language'
df6['Translated Language'] = df6['Translated Language'].replace({
    'ENGLISH': 'English',
    'E': 'English'
}).fillna('Not Specified')
df6['Translated Language'] = df6['Translated Language'].str.strip()

# Normalize 'Translation Quality'
df6['Translation Quality'] = df6['Translation Quality'].replace({
    'Correct': 'Good', # Added this mapping
    'good': 'Good',
    'correct': 'Good',
    'GOOD': 'Good',
    'CORRECT': 'Good',
    'Major Error': 'Poor', # Added this mapping
    'Minor Error': 'Poor', # Added this mapping
    'Incorrect': 'Poor',
    'Wrong': 'Poor',
    'Not able trnaslate disease word in telugu(తెగుళ్ల)': 'Poor',
    'N A': 'Not Applicable',
    'NIL': 'Not Applicable'
}).fillna('Not Applicable')
df6['Translation Quality'] = df6['Translation Quality'].str.strip()

# Normalize 'Tagging'
df6['Tagging'] = df6['Tagging'].replace({
    'Correctly tagged as duplicate': 'Correctly Tagged as Duplicate',
    'Dynamic but not tagged': 'Not Tagged',
    'Duplicate but not tagged': 'Not Tagged',
    'Wrongly Tagged as Duplicate': 'Wrongly Tagged',
    'Wrongly Tagged as Dynamic': 'Wrongly Tagged',
    'DYNAMIC': 'Correctly Tagged as Dynamic'
}).fillna('Not Applicable')
df6['Tagging'] = df6['Tagging'].str.strip()

# Normalize 'Allocated to Reviewer?'
# First, identify reviewer names. Assuming any string not explicitly 'Yes', 'No', 'NIL', 'na', 'Not Applicable' is a name.
reviewer_names = df6['Allocated to Reviewer?'].unique()
reviewer_names_to_yes = [name for name in reviewer_names if isinstance(name, str) and name not in ['Yes', 'No', 'NIL', 'na', 'NO', 'yes', 'YES', 'Yes ', 'no', 'NO ', 'Not yet', 'Correct', 'Successfully Identified as Duplicate/Dynamic', 'Correctly Tagged as Duplicate', 'Wrongly Identified as Duplicate/Dynamic', 'Wrongly Tagged as Duplicate', 'Duplicate', 'Duplicate but not tagged']]

df6['Allocated to Reviewer?'] = df6['Allocated to Reviewer?'].replace({
    'Successfully Identified as Duplicate/Dynamic': 'Yes',
    'Correctly Tagged as Duplicate': 'Yes',
    'Wrongly Identified as Duplicate/Dynamic': 'No',
    'Wrongly Tagged as Duplicate': 'No',
    'Duplicate': 'Yes', # Assuming 'Duplicate' means it was allocated for review as a duplicate
    'Duplicate but not tagged': 'No', # Not allocated if not tagged
    'yes': 'Yes',
    'YES': 'Yes',
    'Yes ': 'Yes',
    'NO': 'No',
    'no': 'No',
    'NO ': 'No',
    'Not yet': 'No',
    'Correct': 'Yes',
    'NIL': 'Not Applicable',
    'na': 'Not Applicable',
    **{name: 'Yes' for name in reviewer_names_to_yes}
}).fillna('Not Applicable')
df6['Allocated to Reviewer?'] = df6['Allocated to Reviewer?'].str.strip()


# Display unique values to confirm normalization
print("Normalized unique values for 'Question in Review Model?':", df6['Question in Review Model?'].unique())
print("Normalized unique values for 'Question Correctly Framed?':", df6['Question Correctly Framed?'].unique())
print("Normalized unique values for 'Original Language':", df6['Original Language'].unique())
print("Normalized unique values for 'Translated Language':", df6['Translated Language'].unique())
print("Normalized unique values for 'Translation Quality':", df6['Translation Quality'].unique())
print("Normalized unique values for 'Tagging':", df6['Tagging'].unique())
print("Normalized unique values for 'Allocated to Reviewer?':", df6['Allocated to Reviewer?'].unique())
```
